```python
# identify all compartments involved and build src, dest tables
# identify which table indices are involved in the adjustment
# apply the adjustment to the flow array (whose indices are the above table)
# return the adjusted (whole flow) array, with index

def comps_involved(query):
    src_strat, src_trait_list = query
    strat_idx = strats.index(src_strat)
    comp_idx = np.where(np.isin(all_comp_table[strat_idx], src_trait_list))[0]
    return comp_idx

in_comps = []
out_comps = []
flow_idx = []

for source_query, dest_query in pop_movement.flow_cat_pairs:
    source_comps = comps_involved(source_query)
    dest_comps = comps_involved(dest_query)
    in_comps.append(source_comps)
    out_comps.append(dest_comps)

src_idx, dest_idx = np.concat(in_comps), np.concat(out_comps)

comp_vals = np.linspace(1.0,ncomps,ncomps)


comps_involved(("age",["young"]))

from jax import numpy as jnp

comp_vals = jnp.array(comp_vals)
        
```
